```python
import pandas as pd
import numpy as np

print("NOMINAL DATASET")

data = pd.DataFrame({
    'Color': ['Red', 'Red', 'Blue', 'Green'],
    'Shape': ['Circle', 'Square', 'Circle', 'Triangle']
})

print(data)

n = len(data)
matrix = np.zeros((n, n))

for i in range(n):
    for j in range(n):
        count = 0

        for col in data.columns:
            if data.loc[i, col] != data.loc[j, col]:
                count += 1

        matrix[i, j] = count / len(data.columns)

print("\nDISSIMILARITY MATRIX - NOMINAL")

print(pd.DataFrame(
    matrix,
    index=['01', '02', '03', '04'],
    columns=['01', '02', '03', '04']
).round(2))
```

```python
import pandas as pd
import numpy as np
data=pd.DataFrame({
    'Student': [1,1,0,0],
    'Employed':[0,1,1,0],
    'Purchased':[1,1,0,1]
})
print("BINARY DATASET")
print(data)
n=len(data)
matrix=np.zeros((n, n))
for i in range(n):
  for j in range(n):
    diff=0
    for col in data.columns:
      if data.loc[i,col]!=data.loc[j,col]:
        diff +=1
    matrix[i,j]=diff/len(data.columns)
print("\nDISSIMILARITY MATRIX - BINARY")

print(pd.DataFrame(
    matrix,
    index=['01', '02', '03', '04'],
    columns=['01', '02', '03', '04']
).round(2))
```

```python
import pandas as pd
import numpy as np
data = pd.DataFrame({
    'Satisfaction':['Low','Medium','High','Medium']
})

print("ORDINAL DATASET")
print(data)
rank={
    'Low':1,
    'Medium':2,
    'High':3
}

n=len(data)
matrix=np.zeros((n, n))

max_rank = max(rank.values())
min_rank = min(rank.values())
max_rank_diff = max_rank - min_rank

for i in range(n):
  for j in range(n):
     x=rank[data.loc[i,'Satisfaction']]
     y=rank[data.loc[j,'Satisfaction']]
     if max_rank_diff > 0:
        matrix[i,j]=abs(x-y)/max_rank_diff
     else:
        matrix[i,j]=0

print("\nDISSIMILARITY MATRIX - ORDINAL")
print(pd.DataFrame(
    matrix,
    index=[f'{k+1:02}' for k in range(n)],
    columns=[f'{k+1:02}' for k in range(n)]
).round(2))
```
